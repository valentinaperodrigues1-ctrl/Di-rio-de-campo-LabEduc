```python

hotels = pd.read_json('/hotels.jsonl', lines=True)
room_types = pd.read_json('/room_types.jsonl', lines=True)
daily_market = pd.read_json('/daily_market.jsonl', lines=True)
weekly_performance = pd.read_json('/weekly_performance.jsonl', lines=True)
rate_decisions = pd.read_json('/rate_decisions.jsonl', lines=True)
reviews = pd.read_json('/reviews.jsonl', lines=True)
review_scenarios = pd.read_json('/review_scenarios.jsonl', lines=True)

tabelas = {
    "hotels": hotels,
    "room_types": room_types,
    "daily_market": daily_market,
    "weekly_performance": weekly_performance,
    "rate_decisions": rate_decisions,
    "reviews": reviews,
    "review_scenarios": review_scenarios
}

for nome, tabela in tabelas.items():
    print(nome, ":", tabela.shape[0], "linhas e", tabela.shape[1], "colunas")
```

### **Encontre a coluna do cadastro de hotéis que guarda uma lista dentro de um único campo. Pense em como transformá-la para responder, por exemplo, "quantos hotéis têm restaurante?".**

No hotels.jsonl. a coluna amenities(comodidades), tem uma lista de comodidades que cada hotel tem.

```python

quantidader = hotels["amenities"].apply(
    lambda lista: "restaurant" in lista if isinstance(lista, list) else False
).sum()

print("Hotéis com restaurante:", quantidader)
```

```python

quantidadea = hotels["amenities"].apply(
    lambda lista: "gym" in lista if isinstance(lista, list) else False
).sum()

print("Hotéis com academia:", quantidadea)
```
